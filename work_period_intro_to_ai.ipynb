```python
from sklearn.datasets import load_wine
from sklearn.model_selection import KFold
import numpy as np

# Load Wine dataset
wine = load_wine()
X = wine.data
y = wine.target
n_samples = X.shape[0]
indices = np.arange(n_samples)

print("PART 1: Wine dataset with 10 folds using sklearn KFold")
print("=" * 65)

kfold_10 = KFold(n_splits=10, shuffle=True, random_state=42)

for fold_number, (train_idx, test_idx) in enumerate(kfold_10.split(X), start=1):
    print(f"\nFold {fold_number}")
    print(f"Train indices ({len(train_idx)} samples): {train_idx[:5]}")
    print(f"Test indices  ({len(test_idx)} samples): {test_idx[:5]}")
```

```python
# PART 2: Manual 5-fold split
print("\n\nPART 2: Wine dataset with 5 folds prepared manually")
print("=" * 60)

# Keep the same shuffled order for reproducibility
rng = np.random.default_rng(42)
shuffled_indices = rng.permutation(n_samples)

# Split into 5 folds as evenly as possible
manual_folds = np.array_split(shuffled_indices, 5)

for fold_number in range(5):
    test_idx = manual_folds[fold_number]
    train_parts = [manual_folds[i] for i in range(5) if i != fold_number]
    train_idx = np.concatenate(train_parts)

    print(f"\nFold {fold_number + 1}")
    print(f"Train indices ({len(train_idx)} samples): {train_idx[:5]}")
    print(f"Test indices  ({len(test_idx)} samples): {test_idx[:5]}")
```
